```python
import pandas as pd
```

```python
df = pd.read_csv('../Dataset/new_documents.csv')
```

```python
print(df.shape, '\n')
df.head()
```

```python
title = 'xxxxxxxxxxx'
text = 'lksdajf jlksajf s jlasj f;llj kjsdal;jf sdljkk sssssssssssssssssssssssssss'
url = 'No url'
authors = "['Abdul Rehman', 'Tahir']"
timestamp = '2020-07-01 00:00:00'
tags = "['Test', 'Sample']"

text_length = len(text.split())
title_length = len(title.split())
num_tags = len(eval(tags))
num_authors = len(eval(authors))

print(f'Text Length: {text_length}\nTitle Length: {title_length}\nNumber of Tags: {num_tags}\nNumber of Authors: {num_authors}')
```

```python
data = {
    'title': [title],
    'text': [text],
    'url': [url],
    'authors': [authors],
    'timestamp': [timestamp],
    'tags': [tags],
    'text_length': [text_length],
    'title_length': [title_length],
    'num_tags': [num_tags],
    'num_authors': [num_authors]
}
```

```python
df_new_doc = pd.DataFrame(data)
df = pd.concat([df, df_new_doc], ignore_index=True)
```

```python
# Saving this in the same file
df.to_csv('../Dataset/new_documents.csv', index=False)
print('New Document Added Successfully!')
```
