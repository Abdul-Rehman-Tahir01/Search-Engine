```python
import json
```

```python
# Loading the inverted index in the RAM
with open('inverted_index.json') as f:
    inverted_index = json.load(f)
```

```python
# Initializing barrels
barrels = {
    'a_c': {},
    'd_h': {},
    'i_m': {},
    'n_r': {},
    's_z': {},
}
```

```python
# Making barrels on the basis of the first letter of the words in the inverted index
for word, data in inverted_index.items():
    if 'a' <= word[0] <= 'c':
        barrels['a_c'][word] = data

    elif 'd' <= word[0] <= 'h':
        barrels['d_h'][word] = data

    elif 'i' <= word[0] <= 'm':
        barrels['i_m'][word] = data

    elif 'n' <= word[0] <= 'r':
        barrels['n_r'][word] = data

    elif 's' <= word[0] <= 'z':
        barrels['s_z'][word] = data
```

```python
# Saving each barrel to a JSON file
for barrel_name, barrel_data in barrels.items():
    with open(f'Barrels/barrel_{barrel_name}.json', 'w') as f:
        json.dump(barrel_data, f)

print('Inverted index has been divided into barrels successfully.')
```
