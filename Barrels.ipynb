```python
import json
```

```python
# Loading the inverted index in the RAM
with open('inverted_index.json') as f:
    inverted_index = json.load(f)

print(f'Inverted index loaded with {len(inverted_index)} words')
```

```python
# Setting parameters
barrel_size = 10000
num_barrels = (len(inverted_index) + barrel_size - 1) // barrel_size

print(f'The total number of barrels made will be {num_barrels}')
```

```python
inverted_index_items = list(inverted_index.items())
```

```python
# Processing and saving each barrel separately
for i in range (num_barrels):
    barrel = inverted_index_items[i * barrel_size : (i + 1) * barrel_size]
    barrel_dict = dict(barrel)

    # Saving the current barrel to JSON file
    print(f'Saving the barrel {i+1} with {len(barrel_dict)} word ids.')
    with open(f'Barrels/barrel_{i+1}.json', 'w') as f:
        json.dump(barrel_dict, f)
    
    print(f'Barrel {i+1} saved successfully.')

    del barrel_dict
```
