```python
# Importing Libraries
import json
```

```python
# Loading the forward index file
with open('forward_index.json') as f:
    forward_index = json.load(f)

print(f'Forward Index loaded with {len(forward_index)} documents')
```

```python
# Initializing inverted index
inverted_index = {}
```

```python
# Build the inverted index
for doc_id, tokens in forward_index.items():
    for token in tokens:
        if token not in inverted_index:
            inverted_index[token] = {"df": 0, "postings": {}}  # Initializing the inverted index dictionary with document frequency and postings
        if doc_id not in inverted_index[token]["postings"]:
            inverted_index[token]["postings"][doc_id] = 0  # Initializing the term frequency
            inverted_index[token]["df"] += 1  # Increment document frequency
        inverted_index[token]["postings"][doc_id] +=1  # Incrementing the term frequency
```

```python
# Sort document ids in postings
for token, data in inverted_index.items():
    sorted_postings = {
        doc_id: tf for doc_id, tf in sorted(data["postings"].items(), key=lambda x: int(x[0][3:]))
    }
    inverted_index[token]["postings"] = sorted_postings
    
```

```python
# Save the inverted index to a JSON file
with open('inverted_index.json', 'w') as f:
    json.dump(inverted_index, f)

print(f'Inverted index saved with {len(inverted_index)} terms.')
```
