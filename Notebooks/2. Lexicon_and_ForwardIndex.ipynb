```python
# Importing Libraries
import pandas as pd
import nltk
from nltk import pos_tag
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords, wordnet
from nltk.stem import WordNetLemmatizer
import string
import json
```

```python
# Donwload the necessary NLTK files
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
# Loading the preprocessed dataset
articlesPreprocessed_dataset = pd.read_csv('../datasets/medium_articles_preprocessed.csv')
```

```python
# Printing the shape and head of the dataset
print(articlesPreprocessed_dataset.shape, '\n')
articlesPreprocessed_dataset.head()
```

```python
# Creating the list of stopwords to be removed from the text
stop_words = set(stopwords.words('english'))
stop_words
```

```python
# Initializing the WordNetLemmatizer and SpellChecker
lemmatizer = WordNetLemmatizer()
```

```python
# Checking the punctuation marks
string.punctuation
```

<hr>

```python
# Function to convert NLTK POS tags to WordNet POS tags
def get_wordnet_pos(tag):
    if tag.startswith('J'):  # Adjective
        return wordnet.ADJ
    elif tag.startswith('V'):  # Verb
        return wordnet.VERB
    elif tag.startswith('N'):  # Noun
        return wordnet.NOUN
    elif tag.startswith('R'):  # Adverb
        return wordnet.ADV
    else:
        return None  # Other POS
```

```python
# Function to preprocess the batch of text
def preprocess(title_text_pairs, doc_ids, lexicon):
    forward_index_batch = {}
    
    for (title, text), doc_id in zip(title_text_pairs, doc_ids): 
        # Tokenization and normalization
        title_tokens = word_tokenize(title.lower())
        text_tokens = word_tokenize(text.lower())

        # Lemmatization and stopword removal
        title_tokens = [
            lemmatizer.lemmatize(word, pos=get_wordnet_pos(pos_tag([word])[0][1]) or 'n')
            for word in title_tokens if word.isalpha()
            and word not in stop_words 
            and word not in string.punctuation
        ]
        
        text_tokens = [
            lemmatizer.lemmatize(word, pos=get_wordnet_pos(pos_tag([word])[0][1]) or 'n')
            for word in text_tokens if word.isalpha()
            and word not in stop_words 
            and word not in string.punctuation
        ]

        # Mapping words to word IDs
        token_ids_title = []
        for word in title_tokens:
            if word not in lexicon:
                lexicon[word] = len(lexicon)  # Add new word to lexicon with unique ID
            token_ids_title.append(lexicon[word])

        token_ids_text = []
        for word in text_tokens:
            if word not in lexicon:
                lexicon[word] = len(lexicon)  # Add new word to lexicon with unique ID
            token_ids_text.append(lexicon[word])

        # Updating forward index with title and text word IDs
        forward_index_batch[doc_id] = {
            'title': token_ids_title,
            'text': token_ids_text
        }

    return forward_index_batch, lexicon
```

```python
# Initializing the lexicon and forward index
lexicon = {}
forward_index = {}

batch_size = 5000
doc_id_prefix = "doc"

for i in range(0, len(articlesPreprocessed_dataset), batch_size):
    # Get batch texts and document ids (separate title and text)
    batch_titles = articlesPreprocessed_dataset['title'][i:i+batch_size]
    batch_texts = articlesPreprocessed_dataset['text'][i:i+batch_size]
    batch_texts_titles = list(zip(batch_titles, batch_texts))
    
    doc_ids = [f"{doc_id_prefix}{j}" for j in range(i, i + len(batch_texts_titles))]

    # Preprocessing the batch with document ids and lexicon
    batch_forward_index, lexicon = preprocess(batch_texts_titles, doc_ids, lexicon)

    # Updating the forward index and lexicon
    forward_index.update(batch_forward_index)
```

```python
# Saving lexicon to a JSON file
with open('lexicon.json', 'w') as f:
    json.dump(lexicon, f, default=str)

print(f'Lexicon created with {len(lexicon)} unique terms.')

# Saving forward index to a JSON file
with open('forward_index.json', 'w') as f:
    json.dump(forward_index, f)

print(f'Forward index created with {len(forward_index)} documents.')
```

```python
# Reading the created lexicon from the JSON file
with open('lexicon.json', 'r') as f:
    lexicon_file = json.load(f)

print(f'Lexicon loaded with {len(lexicon_file)} terms.')
list(lexicon_file.items())[:10]
```
