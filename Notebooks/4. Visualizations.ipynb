```python
import matplotlib.pyplot as plt
import numpy as np
from collections import defaultdict
from collections import Counter
import json
```

```python
with open('../JSON Files/lexicon.json') as f:
    lexicon = json.load(f)
```

```python
with open('../JSON Files/inverted_index.json') as f:
    inverted_index = json.load(f)
```

```python
# Initialize a dictionary to count words starting with each letter
letter_counts = defaultdict(int)

# Iterate over the lexicon and count the first letter of each word
for word in lexicon:
    first_letter = word[0].lower()  # Convert to lowercase to handle case sensitivity
    if first_letter.isalpha():  # Ensure the letter is alphabetic
        letter_counts[first_letter] += 1

# Extract the letters (A-Z) and their corresponding counts
letters = list("abcdefghijklmnopqrstuvwxyz")
counts = [letter_counts[letter] for letter in letters]

# Plot the results
plt.figure(figsize=(10, 5))
plt.bar(letters, counts, color='green')
plt.xlabel('Starting Letter')
plt.ylabel('Frequency')
plt.title('Alphabetical Frequency Distribution')
plt.show()
```

```python
# Calculate word lengths
word_lengths = [len(word) for word in lexicon]
length_counts = Counter(word_lengths)

# Sort by word length
lengths = sorted(length_counts.keys())
counts = [length_counts[length] for length in lengths]

# Create a bar chart
plt.figure(figsize=(8, 5))
plt.bar(lengths, counts, color='salmon')
plt.xlabel('Word Length')
plt.ylabel('Frequency')
plt.title('Word Length-Based Frequency Distribution')
plt.xticks(lengths)
plt.show()
```
