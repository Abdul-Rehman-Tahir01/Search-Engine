```python
import json
```

```python
# Loading the inverted index in the RAM
with open('../JSON Files/Barrels/barrel_3.json') as f:
    barrel = json.load(f)

print(f'Barrel_3 loaded with {len(barrel)} words')
```

```python
# Setting parameters
barrel_size = 250
num_barrels = (len(barrel) + barrel_size - 1) // barrel_size

print(f'The total number of barrels made will be {num_barrels}')
```

```python
barrel_items = list(barrel.items())
```

```python
# Processing and saving each barrel separately
for i in range (num_barrels):
    barrel = barrel_items[i * barrel_size : (i + 1) * barrel_size]
    barrel_dict = dict(barrel)

    # Saving the current barrel to JSON file
    print(f'Saving the barrel_250_{i+1+40} with {len(barrel_dict)} word ids.')
    with open(f'../JSON Files/Barrels/New Barrels/barrel_250_{i+1+40}.json', 'w') as f:
        json.dump(barrel_dict, f)
    
    print(f'Barrel_250_{i+1+40} saved successfully.')

    del barrel_dict
```
