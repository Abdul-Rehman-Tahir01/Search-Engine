```python
import pandas as pd
import json
```

```python
df = pd.read_csv('../datasets/medium_articles_preprocessed.csv')
```

```python
print(df.shape, '\n')
df.head()
```

```python
metadata = {}
for index, row in df.iterrows():
    doc_id = f'doc{index}'
    metadata[doc_id] = {
        "title": row['title'],
        "url": row['url'],
        "authors": row['authors'],
        "tags": row['tags']
    }

print(dict(list(metadata.items())[:5]))
```

```python
with open('metadata.json', 'w') as f:
    json.dump(metadata, f)
```
