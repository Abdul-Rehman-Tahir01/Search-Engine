```python
import json
```

```python
# Loading the lexicon in the RAM
with open('lexicon.json', 'r') as f:
    lexicon = json.load(f)
```

```python
# Single word query search
def single_word_search(query, lexicon):
    query = query.lower()  # Normalizing the query
    
    if query not in lexicon:
        return f"Word '{query}' not found in the lexicon"
    
    # Fetch the posting list from the inverted index file
    with open('inverted_index.json', 'r') as f:
        for line in f:
            inverted_index_line = json.loads(line)

            if query in inverted_index_line:
                return inverted_index_line[query]
            
    return f"Word '{query}' not found in the inverted index"
```

```python
result = single_word_search('universe', lexicon)
print(result)

# Mental: 10973, Health: 21560
```
